Analise exploratoria: Sempre Negocios (tem GPS via JSON-LD!).

```python
import datawork
datawork.setup()
```

```python
from datawork.io.loaders import load_scraping_results
from datawork.profiling import completeness_report, value_distribution
from datawork.display import show_sample

FONTE_ID = ""  # Preencher com UUID
df = load_scraping_results(FONTE_ID)
print(f"{len(df)} registros, {len(df.columns)} colunas")
df.columns.tolist()
```

```python
completeness_report(df)
```

```python
show_sample(df, n=5)
```

```python
# WordPress + Houzez tem GPS via JSON-LD
if 'latitude' in df.columns:
    coords = df[['latitude', 'longitude']].dropna()
    print(f"{len(coords)}/{len(df)} registros com GPS")
else:
    print("Sem coluna latitude — verificar parsing do JSON-LD")
```

```python
for col in ["city", "area_total", "price_text", "cep"]:
    if col in df.columns:
        print(f"\n--- {col} ---")
        display(value_distribution(df, col))
```
