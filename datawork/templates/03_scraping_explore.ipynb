```python
import datawork
datawork.setup()
```

```python
from datawork.io.loaders import load_scraping_results
from datawork.profiling import completeness_report, value_distribution
from datawork.display import show_sample, show_stats

FONTE_ID = ""  # UUID da fonte de scraping
df_raw = load_scraping_results(FONTE_ID)
print(f"Loaded {len(df_raw)} records with {len(df_raw.columns)} columns")
df_raw.columns.tolist()
```

```python
# Completude: quais campos tem dados?
completeness_report(df_raw)
```

```python
# Amostra dos dados
show_sample(df_raw, n=5)
```

```python
# Distribuicao de valores por campo chave
for col in ["city", "type", "tipo_operacao"]:
    if col in df_raw.columns:
        print(f"\n--- {col} ---")
        display(value_distribution(df_raw, col))
```

```python
# Analisar campos vs CanonicalRecord
from petrus.domain.entities.mdm_types import CanonicalRecord
canonical_fields = set(CanonicalRecord.__dataclass_fields__.keys())
raw_fields = set(df_raw.columns)
print("Campos no raw que mapeiam para CanonicalRecord:")
print(raw_fields & canonical_fields)
print("\nCampos no raw sem mapeamento direto:")
print(raw_fields - canonical_fields - {"_registro_id", "_created_at", "url"})
```
